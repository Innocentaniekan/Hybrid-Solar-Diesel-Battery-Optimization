```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
solar = pd.read_csv(
    "Energy_Project/data/abuja_solar_2025.csv",
    skiprows=9
)
```

```python
solar.head()
```

```python
system_size = 5 #KW
```

```python
solar["Solar_Output"] = (
    solar["ALLSKY_SFC_SW_DWN"] * system_size
)
```

```python
solar.head()
```

```python
solar["Load_kWh"] = 40 #KW/day
```

```python
#Energy gap (Demand - Solar)
solar["Energy_Gap"] = (
    solar["Load_kWh"] - solar["Solar_Output"]
)

solar[
    ["Solar_Output",
     "Load_kWh",
     "Energy_Gap"]
].head()
```
